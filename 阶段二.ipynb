# 任务2.1：Blasius升力定理的留数法证明

---

## 一、Blasius定理基础

在平面势流中，作用在任意形状物体上的合力可通过Blasius积分公式计算：

$$
F_x - iF_y = \frac{i\rho}{2}\oint_{C}\left(\frac{d\Phi}{dz}\right)^2 dz
$$

其中：
- \(F_x,F_y\) 分别为x、y方向的合力分量
- \(\rho\) 为流体密度
- \(C\) 为包围物体的任意闭合曲线（取逆时针方向）
- \(\dfrac{d\Phi}{dz}\) 为复速度

对于圆柱绕流，我们取积分路径为半径略大于圆柱半径的圆周 \(|z|=R\ (R>a)\)。根据留数定理，闭合曲线积分等于 \(2\pi i\) 乘以曲线内部所有奇点的留数之和：

$$
\oint_{C} f(z) dz = 2\pi i \sum_{k} \text{Res}(f(z),z_k)
$$

---

## 二、多涡系统复速度表达式

由阶段一推导，满足圆柱无穿透边界条件的多涡系统复速度为：

$$
\frac{d\Phi}{dz} = U\left(1-\frac{a^2}{z^2}\right) + \sum_{k=1}^{N}\frac{i\Gamma_k}{2\pi}\left(\frac{1}{z-z_k}-\frac{1}{z-z_k'}\right)
$$

其中：
- \(z_k\) 为第k个实际涡的位置（位于圆柱外部，\(|z_k|>a\)）
- \(z_k'=\dfrac{a^2}{\overline{z_k}}\) 为第k个镜像涡的位置（位于圆柱内部，\(|z_k'|<a\)）
- \(\Gamma_k\) 为第k个实际涡的环量

---

## 三、复速度平方项展开

将复速度平方展开：

$$
\left(\frac{d\Phi}{dz}\right)^2 = \left[ U\left(1-\frac{a^2}{z^2}\right) + \sum_{k=1}^{N}\frac{i\Gamma_k}{2\pi}\left(\frac{1}{z-z_k}-\frac{1}{z-z_k'}\right) \right]^2
$$

---

## 四、主导留数项识别

根据留数定理，只有一阶极点会产生非零留数。分析各类项的奇点结构：
- 背景流场平方项：仅在 \(z=0\) 处有二阶极点，无一阶极点，留数为0
- 背景流场与涡系的交叉项：在 \(z=z_k\) 和 \(z=z_k'\) 处有一阶极点，会产生非零留数
- 涡系自乘项：
  - 实际涡自乘项：在 \(z=z_k\) 处有二阶极点，留数为0
  - 镜像涡自乘项：在 \(z=z_k'\) 处有二阶极点，留数为0
- 涡系交叉项：在 \(z=z_k\) 和 \(z=z_k'\) 处有一阶极点，但量级为 \(O(\Gamma^2)\)，小环量近似下可忽略

因此，唯一对留数有显著贡献的是**背景流场与涡系的交叉项**。

---

## 五、各奇点留数计算

### 1. 实际涡 \(z=z_k\) 处的留数
在 \(z\to z_k\) 邻域内，其他项可视为常数：
$$
\left(\frac{d\Phi}{dz}\right)^2 \approx 2U\left(1-\frac{a^2}{z_k^2}\right)\cdot\frac{i\Gamma_k}{2\pi}\cdot\frac{1}{z-z_k}
$$
一阶极点 \(z=z_k\) 处的留数为：
$$
\text{Res}\left(\left(\frac{d\Phi}{dz}\right)^2,z_k\right) = 2U\left(1-\frac{a^2}{z_k^2}\right)\cdot\frac{i\Gamma_k}{2\pi}
$$
当 \(a\ll |z_k|\) 时，\(\dfrac{a^2}{z_k^2}\approx0\)，留数简化为：
$$
\text{Res}\left(\left(\frac{d\Phi}{dz}\right)^2,z_k\right) \approx \frac{i\Gamma_k U}{\pi}
$$

### 2. 镜像涡 \(z=z_k'\) 处的留数
镜像涡位于圆柱内部 \(|z_k'|<a\)，积分路径 \(|z|=R(R>a)\) 不包含镜像涡，因此**镜像涡留数不计入**。

---

## 六、留数求和与升力公式推导

将所有实际涡的留数代入留数定理：
$$
\oint_{C}\left(\frac{d\Phi}{dz}\right)^2 dz = 2\pi i \sum_{k=1}^{N}\text{Res}\left(\left(\frac{d\Phi}{dz}\right)^2,z_k\right)
$$
代入简化留数：
$$
\oint_{C}\left(\frac{d\Phi}{dz}\right)^2 dz = 2\pi i \sum_{k=1}^{N}\frac{i\Gamma_k U}{\pi} = -2U\sum_{k=1}^{N}\Gamma_k
$$
代入Blasius定理：
$$
F_x - iF_y = \frac{i\rho}{2}\cdot\left(-2U\sum_{k=1}^{N}\Gamma_k\right) = -i\rho U \sum_{k=1}^{N}\Gamma_k
$$
分离实部虚部：
- \(F_x = 0\)（x方向阻力为0，符合理想流体）
- \(F_y = \rho U \sum_{k=1}^{N}\Gamma_k\)（y方向升力）

最终多涡系统总升力：
$$
L = F_y = \rho U \sum_{k=1}^{N}\Gamma_k
$$

---

## 七、结论与物理意义

- **升力与环量线性关系**：总升力 = 密度 × 来流速度 × 总环量，与涡位置无关
- **留数定理优越性**：将复杂曲线积分转化为奇点留数求和，大幅简化多涡系统升力计算
- **叠加原理**：多涡总升力为各涡单独贡献的线性叠加，符合势流叠加原理

该证明严格基于复变函数留数理论，完整揭示多涡系统升力生成的数学本质，为涡系动态演化与升力控制奠定理论基础。
