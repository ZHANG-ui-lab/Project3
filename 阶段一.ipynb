## 2 多涡系统复势建模与验证（任务1.1）

### 2.1 基础流场：无环量圆柱绕流复势
$$
\Phi_0(z)=U\left(z+\frac{a^2}{z}\right)
$$

### 2.2 镜像法原理
第 $k$ 个实际涡 $z_k$ 对应的镜像涡位置：
$$
z_k'=\frac{a^2}{\overline{z_k}}
$$

### 2.3 多涡系统复势完整表达式
$$
\Phi(z)=U\left(z+\frac{a^2}{z}\right)+\sum_{k=1}^{N}\frac{i\Gamma_k}{2\pi}\ln\left(\frac{z-z_k}{z-\dfrac{a^2}{\overline{z_k}}}\cdot\frac{a}{|z_k|}\right)
$$

复速度：
$$
\frac{d\Phi}{dz}=U\left(1-\frac{a^2}{z^2}\right)+\sum_{k=1}^{N}\frac{i\Gamma_k}{2\pi}\left(\frac{1}{z-z_k}-\frac{1}{z-z_k'}\right)
$$

### 2.4 奇点类型与拓扑分析
- 复势 $\Phi(z)$：对数分支点
- 复速度 $\dfrac{d\Phi}{dz}$：一阶单极点
- 实际涡主导流场，镜像涡仅用于满足边界

### 2.5 洛朗展开验证奇点阶数
$$
\left.\frac{d\Phi}{dz}\right|_{z\to z_k}\approx\frac{i\Gamma_k}{2\pi(z-z_k)}+\mathcal{O}(1)
$$

### 2.6 核心代码实现
```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

def complex_potential_multi_vortex(Z, a, U, vortices):
    Phi = U * (Z + a ** 2 / Z)
    for (xv, yv, Gamma_k) in vortices:
        z_v = xv + 1j * yv
        z_mirror = a ** 2 / np.conj(z_v)
        term = (Z - z_v) / (Z - z_mirror) * (a / np.abs(z_v))
        Phi += 1j * Gamma_k / (2 * np.pi) * np.log(term)
    return Phi

def complex_velocity_multi_vortex(Z, a, U, vortices):
    dPhi = U * (1 - a ** 2 / Z ** 2)
    for (xv, yv, Gamma_k) in vortices:
        z_v = xv + 1j * yv
        z_mirror = a ** 2 / np.conj(z_v)
        dPhi += 1j * Gamma_k / (2 * np.pi) * (1 / (Z - z_v) - 1 / (Z - z_mirror))
    return dPhi
